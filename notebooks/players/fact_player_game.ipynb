```python
import os
from pathlib import Path

import nflreadpy as nfl

from ingestion.ducklake import get_connection

# Lake paths in .env are relative to repo root, not notebooks/players/
repo_root = next(
    p
    for p in [Path.cwd(), *Path.cwd().parents]
    if (p / "ingestion" / "ducklake.py").exists()
)
os.chdir(repo_root)


season = nfl.get_current_season()
player_stats = nfl.load_player_stats(seasons=season)

with get_connection() as conn:
    conn.register("player_stats", player_stats)
    lake_cols = {
        row[0] for row in conn.sql("DESCRIBE lake.raw.nfl_player_stats").fetchall()
    }
    source_cols = set(player_stats.columns)
    new_cols = set(source_cols - lake_cols)
    if new_cols:
        print("new cols")
    else:
        print("no new cols")
new_cols
```
